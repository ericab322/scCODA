```python
import pandas as pd
import numpy as np

import pandas as pd
import glob
import re
thresholds = [0.25, 0.5, 0.75, 0.95]

rows = []

for file in glob.glob("driver_metrics_P*_*.csv"):
    match = re.search(r"P(\d+)_(\w+)_", file)
    P = int(match.group(1))
    prior = match.group(2)
    df = pd.read_csv(file)
    df["threshold"] = thresholds[:len(df)]
    df["P"] = P
    df["prior"] = prior
    rows.append(df)
full_df = pd.concat(rows, ignore_index=True)
full_df = full_df[["P", "prior", "threshold", "TP", "FP", "TN", "FN"]]

print(full_df.head())
```
