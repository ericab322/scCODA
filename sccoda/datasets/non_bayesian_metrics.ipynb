```python
import numpy as np
import pandas as pd

alpha = 0.05

for model in ["ALR_ttest", "ALR_wilcoxon", "CLR", "CLR_ttest", "Haber", "TTest"]:
    for P in [1, 10, 19, 1000, 5000]:

        # Load results and truth
        df = pd.read_csv(
            f'/users/ebrown62/scCODA/sccoda/datasets/results/covariate_scaling/{model}/P={P}/{model}.csv'
        )
        true_df = pd.read_csv(
            f'/users/ebrown62/scCODA/sccoda/datasets/scCODA_simulated_N=200_K=5_P={P}_SEED=0_true_beta.csv'
        )

        # Identify true and null effects
        true_effects = []
        null_effects = []

        for col in [c for c in true_df.columns if c.startswith("CT")]:
            effects = true_df[col].to_numpy()
            if np.any(effects != 0):
                true_effects.append(col)
            else:
                null_effects.append(col)

        # Initialize counts
        TP = FP = TN = FN = 0

        # Evaluate
        for col in [c for c in df.columns if c.startswith("CT")]:
            p = df[col].values[0]

            if col in true_effects:
                if p < alpha:
                    TP += 1
                else:
                    FN += 1
            else:  # null effect
                if p < alpha:
                    FP += 1
                else:
                    TN += 1

        # Save CSV
        results = pd.DataFrame({
            "TP": [TP],
            "TN": [TN],
            "FP": [FP],
            "FN": [FN]
        })

        results.to_csv(f"metrics_P={P}_{model}.csv", index=False)
        print(f"{model}, P={P}: TP={TP}, TN={TN}, FP={FP}, FN={FN}")
```
